# jupyterlab-diff demo

This notebook demonstrates the [jupyterlab-diff](https://github.com/jupyter-ai-contrib/jupyterlab-diff) extension, which provides commands for showing cell and file diffs with multiple diffing strategies.

The buttons below use the JupyterLab *command linker*, so clicking them runs the `jupyterlab-diff` commands directly — no code execution needed. The last section shows how to trigger the same commands programmatically.

## Unified cell diff

Show the changes inline in the cell editor, with added and removed lines marked. Individual chunks can be accepted or rejected.

<button class="jp-Button jp-mod-styled jp-mod-accept" data-commandlinker-command="jupyterlab-diff:unified-cell-diff" data-commandlinker-args='{"cellId": "unified-diff-cell", "originalSource": "def greet(name):\n    print(\"Hello \" + name)", "newSource": "def greet(name, punctuation=\"!\"):\n    message = f\"Hello {name}{punctuation}\"\n    print(message)\n    return message", "showActionButtons": true}'>Show unified diff on the cell below</button>

With `allowInlineDiffs: true`, chunks that keep the same number of lines and only change a small part of a line are rendered *within* the line — deleted text struck through next to the added text — instead of as separate removed/added line blocks. The button below uses a small single-line edit to show this:

<button class="jp-Button jp-mod-styled jp-mod-accept" data-commandlinker-command="jupyterlab-diff:unified-cell-diff" data-commandlinker-args='{"cellId": "unified-diff-cell", "originalSource": "def greet(name):\n    print(\"Hello \" + name)", "newSource": "def greet(name):\n    print(\"Howdy there, \" + name + \"!\")", "showActionButtons": true, "allowInlineDiffs": true}'>Show unified diff (with inline diffs)</button>

```python
def greet(name):
    print("Hello " + name)
```

## Programmatic usage

The commands can also be executed programmatically. This site is configured with `exposeAppInBrowser`, so the JupyterLab application object is available as `window.jupyterapp` — in the browser dev tools console, or from a notebook with a small helper like the one below. Run the next two cells to show a diff on the last cell.

```python
import json
from IPython.display import Javascript, display


def execute_command(command, **args):
    """Execute a JupyterLab command from the kernel."""
    display(Javascript(
        f'window.jupyterapp.commands.execute({json.dumps(command)}, {json.dumps(args)});'
    ))
```

```python
execute_command(
    'jupyterlab-diff:split-cell-diff',
    cellId='playground-cell',
    originalSource='total = 0\nfor i in range(10):\n    total += i\nprint(total)',
    newSource='total = sum(range(10))\nprint(total)',
)
```

```python
total = 0
for i in range(10):
    total += i
print(total)
```
